```python
import csv
import struct
c = "earthquake_dataset.csv"

MAX_COUNTRY_LEN = 32

with open(c, "r") as csvfile, open("data.dat", "wb") as datfile:
    reader = csv.DictReader(csvfile)
    for row in reader:
        longitude = float(row["Longitude"])
        country = row["Country"].encode("utf-8")[:MAX_COUNTRY_LEN]  # truncate if too long
        country += b"\x00" * (MAX_COUNTRY_LEN - len(country))        # pad if too short
        datfile.write(struct.pack("f32s", longitude, country))
```
